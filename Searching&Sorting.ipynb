```python
def linearsearch(a,T):
    for i in range(len(a)):
        if a[i] == T:
            return i
    return -1
print(linearsearch([10, 20, 30, 40, 50], 30))

```

```python
def binarysearch(a,T):
    low = 0
    high = len(a) - 1
    while low <= high:
        mid = (low + high) // 2
        if a[mid] == T:
            return mid
        elif a[mid] < T:
            low = mid + 1
        else:
            high = mid - 1
    return -1
print(binarysearch([10, 20, 30, 40, 50, 60],40))

```

```python
def bubblesort(a):
    for i in range(len(a)-2):
        for j in range(i+1,len(a)-1):
            if a[i] > a[j]:
                a[i], a[j] = a[j], a[i]
                print(a)
bubblesort([5,2,8,1,3])
```
